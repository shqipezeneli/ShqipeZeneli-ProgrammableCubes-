```python
import pandas as pd
import numpy as np

print("--- Projekt: Programmable Cubes ---")
print("Algoritmet e Kërkimit të Rrugës (Pathfinding)")

# 1. Definimi i gjendjes fillestare dhe gjendjes synim për Kubat
start_pos = (0, 0, 0)
target_pos = (10, 5, 2)

# 2. Algoritmi i thjeshtë i lëvizjes (Greedy Search Baseline)
path = []
current_pos = list(start_pos)

while current_pos != list(target_pos):
    path.append(tuple(current_pos))
    for i in range(3):
        if current_pos[i] < target_pos[i]:
            current_pos[i] += 1
            break

path.append(tuple(current_pos))

print(f"Hapat e lëvizjes u llogaritën! Gjithsej hapa: {len(path) - 1}")

# 3. Ruajtja e rezultatit për Submission
df = pd.DataFrame(path, columns=['X', 'Y', 'Z'])
df.to_csv('submission.csv', index=False)
print("Skedari 'submission.csv' u krijua me sukses te dosja /kaggle/working!")
```
